```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'app.py').exists() else Path.cwd().parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
import pandas as pd
from src.utils import format_inr
from database.database import read, connect

def show(frame):
    view = frame.copy()
    for column in view:
        if column.endswith('_inr'):
            view[column] = view[column].map(lambda value: format_inr(value, 2))
    print(view.to_string(index=False))
```

```python
m = pd.read_csv(ROOT / 'reports/forecast_metrics.csv')
show(m[m.dimension == 'total'])
f = pd.read_csv(ROOT / 'reports/forecasts.csv')
show(f[f.dimension == 'total'])
assert (pd.to_datetime(m.validation_end) < pd.to_datetime(m.test_start)).all()
```
