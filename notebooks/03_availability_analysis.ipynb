# Availability and lost-demand estimates

Where is unfulfilled demand concentrated?

All data is synthetic. Run the pipeline first. These notebooks consume the same executed reports and canonical definitions as the application. Monetary values are INR.


```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'app.py').exists() else Path.cwd().parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
import pandas as pd
from src.utils import format_inr
from database.database import read, connect

def show(frame):
    view = frame.copy()
    for column in view:
        if column.endswith('_inr'):
            view[column] = view[column].map(lambda value: format_inr(value, 2))
    print(view.to_string(index=False))
```

```python
a = pd.read_csv(ROOT / 'reports/availability_metrics.csv')
show(a[['store_id','fill_rate','quantity_unfulfilled','estimated_lost_revenue_inr']].sort_values('estimated_lost_revenue_inr',ascending=False))
assert a.quantity_fulfilled.sum() + a.quantity_unfulfilled.sum() == a.quantity_ordered.sum()
```
